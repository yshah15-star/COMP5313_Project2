## Step 1: Load the data

The dataset was downloaded from Hugging Face as two Parquet files and saved in the `data/` folder:

- `train-00000-of-00001.parquet`: 25,000 reviews used for training
- `test-00000-of-00001.parquet`: 25,000 reviews used for testing

Each row has two columns: `text` (the review) and `label` (`0` = negative, `1` = positive).

```python
import pandas as pd
import re

DATA_DIR = "data"
TRAIN_FILE = f"{DATA_DIR}/train-00000-of-00001.parquet"
TEST_FILE = f"{DATA_DIR}/test-00000-of-00001.parquet"

train_df = pd.read_parquet(TRAIN_FILE)
test_df = pd.read_parquet(TEST_FILE)

pd.set_option('display.max_colwidth', None)

print(train_df.head(1))
```

## Step 2: Clean the text

The raw reviews contain HTML tags (`<br />`), punctuation and numbers.
We clean each review by:

1. Removing HTML tags
2. Keeping only letters (removing numbers and punctuation)
3. Removing extra spaces

The original upper/lower case is kept.

```python


def clean_text(text):
    text = re.sub(r"<.*?>", " ", text)          # remove HTML tags like <br />
    text = re.sub(r"[^a-zA-Z\s]", " ", text)    # keep letters only
    text = re.sub(r"\s+", " ", text).strip()     # remove extra spaces
    return text

train_df["clean_text"] = train_df["text"].apply(clean_text)
test_df["clean_text"] = test_df["text"].apply(clean_text)
```
